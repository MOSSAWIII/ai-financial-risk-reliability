```r
# Petit test de calcul et de manipulation de donnees
rendements <- c(0.012, -0.006, 0.018, 0.004, -0.003)
data.frame(
  moyenne = mean(rendements),
  volatilite = sd(rendements),
  rendement_cumule = prod(1 + rendements) - 1
)
```

```r
# Petit test graphique avec R de base
plot(
  cumsum(rendements),
  type = "b",
  col = "steelblue",
  main = "Rendements cumules (somme)",
  xlab = "Observation",
  ylab = "Rendement cumule"
)
abline(h = 0, lty = 2, col = "grey50")
```
